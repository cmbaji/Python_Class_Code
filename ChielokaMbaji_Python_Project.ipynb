```python
#A system to monitor a person's exercises for the week.
#This code is designed to monitor a person's exercise over a week using dictionaries, lists and functions.
#The code works with defined functions that serve together to build a main function. 
#It uses the main dictionary at the end to tie in the previously defined functions that require user input. 
#The code was originally not so efficient. I added in the .lower() function to help with formatting. 
#I also added an option to show the exercises for the whole week as opposed to just the one day. 

def add_exercise(data):
    day = input("Enter the day of the week: ").lower()
    exercise = input("Enter exercise name: ")
    minutes = int(input("Enter minutes exercised: "))

#The add_exercise() function asks the user to enter the day, the exercise, and how many minutes they exercised.
    
    if day not in data:
        data[day] = []
        
#If the day does not already exist in the dictionary, the program creates a new entry for it. 

    data[day].append((exercise, minutes))
    print("Exercise added.\n")

#The exercise is then added to the list for that day using the .append() method.


def show_all(data):
    if not data:
        print("No exercise data recorded.\n")
        return

    for day in data:
        print(day.capitalize())
        for ex, mins in data[day]:
            print(" ", ex, "-", mins, "minutes")
        print()

#The show_day function works to show the user all of their exercises for a particular day

def weekly_total(data):
    total = 0

    for day in data:
        for ex, mins in data[day]:
            total += mins

    print("Total minutes exercised this week:", total)
    
#This for loop continues to add on minutes to the weekly total as the user works out more and more


def main():
    exercises = {}

#Here, the dictionary named exercises is used to contain the main menu code. It is used in a while loop
#The while loop will break only when the user decides to leave the menu. If they do not choose one of the options
#The code will askk them to go back and pick a valid option
    
    while True:
        print("1. Add Exercise")
        print("2. Show All Exercises")
        print("3. Show Weekly Total")
        print("4. Exit")

        choice = input("Choose an option: ")

        if choice == "1":
            add_exercise(exercises)

        elif choice == "2":
            show_all(exercises)

        elif choice == "3":
            weekly_total(exercises)

        elif choice == "4":
            break

        else:
            print("Invalid option.\n")


main()
```
